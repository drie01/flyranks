## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*


- **Outcome (proxy for "ranks well"): `top10`** = 1 if `avg_position <= 10` (first page of results), else 0. I'll also keep `avg_position` itself as a continuous outcome.
- **Observed, not defined:** `avg_position` is measured by search data, not produced by a rule I wrote. That is what the framing skill asks for.
- **Eligibility:** only pages with enough impressions (`impressions_90d >= 100`) and a non-missing position, because position averaged over a handful of impressions is noise.
- **Proxy limits:** ranking position is not the same as business value (clicks, sessions, conversions). Position is also shaped by things I can't see (backlinks, domain strength). So I will claim only **directional** results, such as "pages in group A are more often on page one".
- **Signals to test (candidates):** `word_count`, `content_age_days`, `days_since_last_update`, `content_type`, `main_intent`, `search_volume`, `competition`, `cpc`, `provider_used` / `model_used`.
- **Do NOT use as signals:** `ctr`, `clicks_*`, `impression_tier`, `position_tier`. They are consequences of position or are derived from it, so using them would be leakage.
- **Confounders to watch:** `client_id` (32 clients, so domain strength differs), and `search_volume` / `competition` (hard keywords rank worse regardless of content).
